```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
pack = pd.read_csv("./pack_HRR.csv", header=None)
body = pd.read_csv("./body_HRR.csv", header=None)
total = pd.read_csv("./total_HRR.csv", header=None)

Data = {
    "Pack": pack,
    "Body": body,
    "Total": total
}
```

```python
for key, data in Data.items():
    data = np.array(data)
    plt.plot(data[:, 0], data[:, 1], label=key)

plt.legend()
```
